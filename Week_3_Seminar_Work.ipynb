```python
# Drop the 'Cabin' column because 77% of its data is missing
titanic_cleaned = titanic.drop(columns=['Cabin'])

# Drop rows that have missing values in the remaining columns (like 'Age' and 'Embarked')
titanic_cleaned = titanic_cleaned.dropna()

# Display the new shape and the first few rows of the cleaned data
print(f"Original shape: {titanic.shape}")
print(f"Cleaned shape: {titanic_cleaned.shape}")
display(titanic_cleaned.head())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
palette = {0: "#d62728", 1: "#2ca02c"}   # red = died, green = survived
```

```python
sns.pairplot(
    data=titanic_cleaned[['Survived', 'Age', 'Fare', 'SibSp', 'Parch']],
    hue='Survived', palette=palette, diag_kind='hist', corner=True,
    plot_kws={'alpha': 0.6}
)
plt.suptitle("Pairplot of Numeric Features by Survival", y=1.02)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.barplot(data=titanic_cleaned, x='Sex', y='Survived', ax=axes[0])
axes[0].set_title("Survival Rate by Sex")
sns.barplot(data=titanic_cleaned, x='Pclass', y='Survived', ax=axes[1])
axes[1].set_title("Survival Rate by Passenger Class")
plt.tight_layout(); plt.show()
```

```python
sns.histplot(data=titanic_cleaned, x='Age', hue='Survived', palette=palette,
             kde=True, bins=30, element='step')
plt.title("Age Distribution by Survival")
plt.show()
```

```python
sns.boxplot(data=titanic_cleaned, x='Pclass', y='Fare', hue='Survived', palette=palette)
plt.yscale('log')   # a few huge fares (~£512) squash the plot otherwise
plt.title("Fare by Class and Survival (log scale)")
plt.show()
```

```python
sns.scatterplot(data=titanic_cleaned, x='Age', y='Fare', hue='Survived',
                style='Sex', palette=palette, alpha=0.7)
plt.yscale('log')
plt.title("Age vs Fare, coloured by Survival")
plt.show()
```

```python
sns.violinplot(data=titanic_cleaned, x='Pclass', y='Age', hue='Sex', split=True)
plt.title("Age Distribution by Class and Sex")
plt.show()
```

```python
sns.catplot(data=titanic_cleaned, x='Sex', y='Survived', col='Pclass',
            kind='bar', height=4)
plt.suptitle("Survival Rate by Sex within each Class", y=1.05)
plt.show()
```

```python
num_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']
sns.heatmap(titanic_cleaned[num_cols].corr(), annot=True, fmt='.2f',
            cmap='coolwarm', center=0)
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.countplot(data=titanic_cleaned, x='Embarked', hue='Survived', palette=palette)
plt.title("Passenger Count by Port of Embarkation and Survival")
plt.show()
```

Sex matters most. About 75% of women survived, compared with about 21% of men.

Class comes next. Survival was about 65% in 1st class, 48% in 2nd and 24% in 3rd.
Fare has a positive link to survival (correlation 0.27). That’s mostly because fare goes with class.

Age on its own has almost no linear link to survival (correlation −0.08). However the histogram still shows a bump of survivors among young children.

Name, Ticket and PassengerId are left out of every chart because they’re identifiers, so plotting them wouldn’t tell you anything.
